```python
#Bibliotecas
import requests
import pandas as pd
```

```python
# Coleta de Dados
# Definir a URL base da requisição
url = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.1/dados?formato=json"

# Executar a requisição
resposta = requests.get(url = url)

# Transformar resposta para tabela
tabela = pd.DataFrame(resposta.json())
tabela
```
