```python
# Instalar e importar bibliotecas
!pip install pygsheets
import pygsheets
import pandas as pd
```

```python
# Autentica com API do Google
api = pygsheets.authorize(client_secret = "api.json")
api
```

```python
# Importar tabela usando a URL
api.open_by_url(
    "https://docs.google.com/spreadsheets/d/19L1EH_Bv2__ydVJvkmL5QMaXc7PS9Vtplh2uZ9EgViU/edit?usp=sharing"
    ).worksheet_by_title("Oceania").get_as_df()
```

```python
# Criar arquivo no Drive
minha_planilha = api.create("orcamento_mensal2")

# Criar folha dentro da planilha
minha_planilha.add_worksheet("folha1")

# Inserir dados na "folha1"
(
   minha_planilha
  .worksheet_by_title("folha1")
  .set_dataframe(
      df = pd.DataFrame({"x": [1, 2, 3], "y": ["a", "b", "c"]}),
      start = (1, 1)
      )
)

# Deletar folha "Sheet1"
minha_planilha.del_worksheet(minha_planilha.worksheet())  # função de dentro retorna a primeira folha e função de fora deleta a mesma

# Importar dados criados no Sheets
api.open_by_url(
    "https://docs.google.com/spreadsheets/d/1oa4ppbQC2nDpXoyts5J0z3Nvj8kRth-AaQJIrlyxG6E/edit?usp=sharing"
    ).worksheet_by_title("folha1").get_as_df()
```
