```python
# Cenário 01 - Criando um banco de dados SQL a partir de arquivos CSV

# Biblioteca
import pandas as pd
```

```python
# Importando arquivo CSV
dados = pd.read_csv(filepath_or_buffer = "https://aluno.analisemacro.com.br/download/45589/?tmstv=1672743605",
                    sep = ";",
                    encoding = "latin_1"
        )
dados
```

```python
# Criar banco de dados SQLite

# Biblioteca
from sqlalchemy import create_engine, inspect
banco_de_dados = create_engine("sqlite:///dados/votacao2022.sqlite")
conexao = banco_de_dados.connect()
conexao
```

```python
# Armazenar dados do CSV como uma tabela a ser criada no banco SQL
dados.to_sql(name = "votacao_acre", con = conexao, index = False)
```

```python
# Cenário 2 - Consultando um banco de dados SQL

# Ver nomes das tabelas
inspecao = inspect(banco_de_dados)
inspecao.get_table_names()
```

```python
# Ver informações das colunas de uma tabela
inspecao.get_columns(table_name = "votacao_acre")
```

```python
# Importar tabela completa do banco SQL
pd.read_sql(sql = "votacao_acre", con = conexao)
```

```python
# Realizar consulta SQL
pd.read_sql(
      sql = "SELECT NM_VOTAVEL, QT_VOTOS, NM_MUNICIPIO, NR_SECAO FROM votacao_acre WHERE NM_MUNICIpIO == 'RIO BRANCO'",
      con = conexao
      )
```

```python
# Encerrar conexão com o banco SQL
conexao.close()
conexao
```
