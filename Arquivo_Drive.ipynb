```python
# Importar biblioteca
from pydrive2.auth import GoogleAuth   # instale a biblioteca se precisar: !pip install pydrive2
from pydrive2.drive import GoogleDrive
```

```python
# Autenticar na API do Google
autenticacao = GoogleAuth()
autenticacao.CommandLineAuth() # acesse o link e siga instruções para obter o código

# Se "conectar" ao Drive
drive = GoogleDrive(autenticacao)
```

```python
# Pesquisar arquivos do Drive e retornar 5 resultados
drive.ListFile({"maxResults": 5}).GetList()[0]
```

```python
# Pesquisar arquivos do Drive por padrão de nome
drive.ListFile({"q": "title contains 'Tendencias'"}).GetList()
```

```python
# Baixar arquivo identificado (id)
drive.ListFile({"q": "title contains 'Tendencias'"}).GetList()[0].GetContentFile("dados.csv")
```

```python
# Envia arquivo local para o Drive
arquivo = drive.CreateFile()
arquivo.SetContentFile("dados.csv")
arquivo.Upload()
```
